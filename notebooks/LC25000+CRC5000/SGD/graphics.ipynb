```python
import re

def extract_losses(log_text: str):
    train_loss = []
    val_loss = []

    pattern = re.compile(
        r"Train Loss:\s*([0-9]*\.?[0-9]+)\s*\|\s*Val Loss:\s*([0-9]*\.?[0-9]+)"
    )

    for line in log_text.splitlines():
        match = pattern.search(line)
        if match:
            train_loss.append(float(match.group(1)))
            val_loss.append(float(match.group(2)))

    return train_loss, val_loss
```

```python
import matplotlib.pyplot as plt

def generate_graph(lossess: str):
    train_losses, val_losses = extract_losses(lossess)

    plt.figure(figsize=(10, 7))
    plt.plot(train_losses, label="Train")
    plt.plot(val_losses, label="Validation")
    plt.title("Loss per Epoch", fontsize=18)
    plt.xlabel("Epoch", fontsize=14)
    plt.ylabel("Loss", fontsize=14)
    plt.legend(fontsize=16)
    plt.xticks(fontsize=16)
    plt.yticks(fontsize=16)
    plt.grid(True)
    plt.show()
```
